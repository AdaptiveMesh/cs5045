## Setup: The Catalog Dataset

**Operation: Represent**: define the library catalog as a list of dictionaries.

Each dictionary represents one book; keys match the CSV column names.
Numeric fields are already integers; `available` is already a boolean.
No file needed: the full 15-book catalog is defined right here.

````python
books = [
    {'title': 'The Signal and the Noise',   'author': 'Nate Silver',           'genre': 'Statistics',           'year': 2012, 'pages': 534, 'available': True},
    {'title': 'Thinking Fast and Slow',      'author': 'Daniel Kahneman',       'genre': 'Psychology',           'year': 2011, 'pages': 499, 'available': False},
    {'title': 'The Hidden Life of Trees',    'author': 'Peter Wohlleben',       'genre': 'Nature',               'year': 2015, 'pages': 288, 'available': True},
    {'title': 'Invisible Women',             'author': 'Caroline Criado Perez', 'genre': 'Social Science',       'year': 2019, 'pages': 410, 'available': True},
    {'title': 'The Art of Statistics',       'author': 'David Spiegelhalter',   'genre': 'Statistics',           'year': 2019, 'pages': 448, 'available': False},
    {'title': 'Sapiens',                     'author': 'Yuval Noah Harari',     'genre': 'History',              'year': 2011, 'pages': 443, 'available': True},
    {'title': 'How Not to Be Wrong',         'author': 'Jordan Ellenberg',      'genre': 'Mathematics',          'year': 2014, 'pages': 468, 'available': False},
    {'title': 'Bad Science',                 'author': 'Ben Goldacre',          'genre': 'Health',               'year': 2008, 'pages': 336, 'available': True},
    {'title': 'The Innovators',              'author': 'Walter Isaacson',       'genre': 'Technology',           'year': 2014, 'pages': 542, 'available': True},
    {'title': 'Data Feminism',               'author': "Catherine D'Ignazio",   'genre': 'Social Science',       'year': 2020, 'pages': 328, 'available': True},
    {'title': 'Nudge',                       'author': 'Richard Thaler',        'genre': 'Behavioral Economics', 'year': 2008, 'pages': 312, 'available': False},
    {'title': 'The Code Book',               'author': 'Simon Singh',           'genre': 'Technology',           'year': 1999, 'pages': 402, 'available': True},
    {'title': 'Factfulness',                 'author': 'Hans Rosling',          'genre': 'Statistics',           'year': 2018, 'pages': 342, 'available': False},
    {'title': 'Weapons of Math Destruction', 'author': "Cathy O'Neil",          'genre': 'Social Science',       'year': 2016, 'pages': 272, 'available': True},
    {'title': 'Range',                       'author': 'David Epstein',         'genre': 'Psychology',           'year': 2019, 'pages': 352, 'available': True},
]

print(f'Catalog has {len(books)} books')
print(f'Keys: {list(books[0].keys())}')
````

---
## Lists: Ordered Sequences

`books` is a **list**. Lists are ordered; indexing starts at 0.

````python
# Total count
print('Total books:', len(books))

# Index: zero-based
print('First book (index 0):', books[0]['title'])
print('Last book  (index -1):', books[-1]['title'])

# Slice: stop index is exclusive
print('\nFirst three titles:')
for b in books[:3]:
    print(f'  {b["title"]}')

# Answer to the pause-and-ask
print('\nbooks[2]["genre"]:', books[2]['genre'])
````

---
## Dictionaries: Named Fields

Each element of `books` is a **dictionary**, one row turned sideways.
Access a field by its name (key), not its position.


````python
# All fields for the first book
print('Fields of books[0]:')
for key, value in books[0].items():
    print(f'  {key:12s} -> {value}')

# Chained access: list index, then dict key
print()
print('Author of books[0]:', books[0]['author'])
print('Pages  of books[1]:', books[1]['pages'])   
print('Available of books[2]:', books[2]['available'])
````

---
## List Comprehensions: Transform in One Line

**Operation: Transform**: produce a new list derived from an existing one.

Pattern: `[expression  for  item  in  collection  if  condition]`



````python
# All titles: loop version
titles_loop = []
for book in books:
    titles_loop.append(book['title'])

# All titles: comprehension version (same result)
titles_comp = [book['title'] for book in books]

print('Same result?', titles_loop == titles_comp)

# Add a condition: only available books
available_titles = [book['title'] for book in books if book['available']]
print(f'\nAvailable ({len(available_titles)} books):')
for t in available_titles:
    print(f'  {t}')

# Answer to pause-and-ask: only titles of books with more than 400 pages
over_400 = [book['title'] for book in books if book['pages'] > 400]
print(f'\nMore than 400 pages ({len(over_400)} books):')
for t in over_400:
    print(f'  {t}')
````

---
## Tuples: Fixed Records

A **tuple** is like a list, but immutable: you cannot change it after creation.
Use a tuple when the record is complete and should not be modified.

Common uses: fixed records `(title, author, year)`, function returning two values.

````python
# A fixed (title, author, year) record
first_record = (books[0]['title'], books[0]['author'], books[0]['year'])
print('Record:', first_record)
print('Type:', type(first_record))
print('Title:', first_record[0])
print('Year :', first_record[2])

# Trying to change: raises TypeError
try:
    first_record[0] = 'something else'
except TypeError as e:
    print(f'\nCannot modify a tuple: {e}')

# A function returning two values as a tuple
def page_stats(book_list):
    return len(book_list), sum(b['pages'] for b in book_list)

count, total = page_stats(books)
print(f'\n{count} books, {total} total pages, {total/count:.0f} avg')
````

---
## Transform: Adding Computed Columns

**Operation: Transform**: create new values from existing ones.

We loop through every book and add a new key to each dictionary.
Existing keys are unchanged; we only add.



````python
# Transform 1: 'decade' derived from 'year'
# Integer division (//) drops the remainder
for book in books:
    book['decade'] = (book['year'] // 10) * 10

print('decade column added. Sample:')
for book in books[:5]:
    print(f'  {book["title"][:35]:35s}  {book["year"]}  ->  {book["decade"]}s')

# Transform 2: 'long' flag (True when pages > 500)
for book in books:
    book['long'] = book['pages'] > 500

print(f'\nbooks[0] now has {len(books[0])} keys: {list(books[0].keys())}')

long_titles = [b['title'] for b in books if b['long']]
print(f'\nBooks over 500 pages ({len(long_titles)}): {long_titles}')
````

---
## Represent: Index and Grouped Dictionaries

**Operation: Represent**: reorganize existing data into a new structure.

Two patterns:
1. **Lookup index:** title -> full book dictionary (instant access, no loop)
2. **Dict of lists:** genre -> list of titles (preview of pandas GroupBy)



````python
# Lookup index: title -> full book dict (one comprehension pass)
catalog_index = {book['title']: book for book in books}

# Instant lookup: no loop needed
book = catalog_index['Factfulness']
print('Factfulness:')
print(f'  Author : {book["author"]}')
print(f'  Year   : {book["year"]}')
print(f'  Decade : {book["decade"]}s')

# Dict of lists: group by genre
by_genre = {}
for b in books:
    g = b['genre']
    if g not in by_genre:
        by_genre[g] = []
    by_genre[g].append(b['title'])

print('\nBooks by genre:')
for genre in sorted(by_genre):
    print(f'  {genre:22s}: {by_genre[genre]}')
````
