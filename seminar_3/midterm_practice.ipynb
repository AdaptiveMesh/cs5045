### P8 (inheritance)

**My prediction** (double-click to edit, write it *before* running the next cell):

...

```python
class Patient:
    def __init__(self, name, sys_bp):
        self.name = name
        self.sys_bp = sys_bp

    def risk_label(self):
        if self.sys_bp > 140:
            return "High Risk"
        return "Standard Risk"

class HighRiskPatient(Patient):
    def __init__(self, name, sys_bp, care_manager):
        super().__init__(name, sys_bp)
        self.care_manager = care_manager

    def risk_label(self):
        return "High Risk (Priority)"

p1 = Patient("Derek", 118)
p2 = HighRiskPatient("Alice", 118, "Dr. Chen")
print(p1.risk_label())
print(p2.risk_label())
print(p2.name, p2.care_manager)
```
