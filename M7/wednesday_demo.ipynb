## Part 1: The Amazon Hiring Tool: Key Facts

Deliver these facts during Segment 2 (5 minutes). State them as facts only.  
Students do the analysis in Segment 3. Do not editorialize until they have weighed in.

---

**The system:**  
Amazon built a machine learning tool to screen resumes for technical roles.  
Development began approximately 2014.  
Goal: rank applicants automatically and reduce recruiter time on initial review.

**The training data:**  
Ten years of resumes submitted to Amazon, labeled with Amazon's historical hiring decisions.  
Technical roles during this period had been predominantly filled by men.

**The label:**  
"Did Amazon hire this person?", meaning: does this resume match the profile of people
Amazon has historically hired?

**The discovered bias:**
- The model penalized resumes containing the word "women's"  
  (examples: "women's chess club captain," "president, women's engineering society").
- The model downgraded graduates of two specific all-women's colleges.
- Amazon engineers removed gender-identifying words from the input.  
  The model continued finding other features correlated with gender.

**The outcome:**  
The tool was shut down internally in 2017.  
Reuters reported on it in October 2018.  
It was never used in production hiring decisions.

**Source:** Jeffrey Dastin, "Amazon scraps secret AI recruiting tool that showed
bias against women," Reuters, October 10, 2018.

## Part 1b: COMPAS Recidivism Scoring: Key Facts

Deliver these during Segment 2, Case 2 (5 minutes). State them as facts only, same as
Amazon. Students do the analysis in Segment 3.

---

**The system:**
COMPAS (Correctional Offender Management Profiling for Alternative Sanctions), a risk
score used by US courts since the 1990s to predict likelihood of reoffending.
Used to help set bail, sentencing, and parole decisions.

**The investigation:**
ProPublica analyzed COMPAS scores for 7,000+ people arrested in Broward County, FL,
comparing predicted risk to actual 2-year reoffense outcomes. Published as
"Machine Bias," May 2016.

**The label:**
"Rearrested within two years," a proxy for "reoffended." Arrest rates reflect policing
patterns as well as behavior.

**The finding (ProPublica's own words):** "the algorithm made mistakes with Black and
white defendants at roughly the same rate, but in very different ways."

| | Black defendants | White defendants |
|---|---|---|
| Falsely flagged high-risk (did not reoffend) | 45% | 23% |
| Falsely flagged low-risk (did reoffend) | 28% | 48% |

**The company's defense:**
Northpointe (COMPAS's maker) showed the tool was equally *calibrated*: a given score
meant roughly the same reoffense probability regardless of race. Both ProPublica's
finding and Northpointe's defense are true at the same time, they measure different
things.

**The outcome:**
COMPAS remains in use today. Wisconsin's Supreme Court upheld its use in sentencing
in *State v. Loomis* (2016), with disclosure requirements, not a ban. Research since
(Chouldechova 2016; Kleinberg, Mullainathan, Raghavan 2016) proved equal calibration
and equal error rates across groups cannot both hold when groups have different
underlying reoffense rates, a mathematical fact, not a fixable bug.

**Source:** Julia Angwin, Jeff Larson, Surya Mattu, Lauren Kirchner, "Machine Bias,"
ProPublica, May 23, 2016.

## Part 2: Five Operations Mapping

Use this during Segment 3, Part A after students name an operation.

| Operation | Role in the Amazon case |
|-----------|------------------------|
| **Represent** | Root cause: training labels encoded past hiring decisions, not job qualifications. The dataset asked "who has Amazon hired?" not "who is qualified?" A decade of biased hiring became the ground truth. |
| **Filter** | Deployment phase: the model's ranked output was filtered, only top-N candidates surfaced to recruiters. The cutoff was a human policy choice with no neutral setting. |
| **Transform** | Amplifier: the model derived implicit scores correlated with gender through proxy features (college names, activity descriptions, writing style) without naming gender. |
| **Aggregate** | Less central: the bias was visible in aggregate (women ranked lower overall) but was discovered through individual case inspection, not group statistics. |
| **Communicate** | Framing made it worse: the tool's output was presented as an "objective ranking," making embedded bias harder to see and easier to trust without scrutiny. |

**Synthesis line for the end of Part A:**  
"We have Represent as the source, Transform as the amplifier, and Communicate as the
packaging that made the problem hard to see. This pattern (source, amplifier, framing)
appears repeatedly in documented failures."

**COMPAS mapping** (the distinctive one here is Aggregate, not Represent):

| Operation | Role in the COMPAS case |
|-----------|------------------------|
| **Represent** | Present, but not the main event: the "rearrested" label is a noisy proxy for "reoffended," shaped by policing patterns. |
| **Transform** | ~137 input features, not race itself, but things correlated with it (neighborhood, employment history, family criminal history). Same proxy mechanism as Amazon, different domain. |
| **Aggregate** | The distinctive answer. A single overall accuracy number (both races predicted about equally well overall) hides the disparity completely. Only disaggregating by group AND by error type (false positive vs. false negative) reveals it. |
| **Communicate** | Northpointe and ProPublica reported different statistics from the same scores and reached opposite conclusions. Which number you choose to report is itself a decision with consequences. |
| **Filter** | Judges use a score threshold to inform bail/sentencing severity, the same cutoff-as-policy-choice issue as Amazon's ranked list. |

**Question 2: The threshold and filter (7 min)**

Target: the threshold is a human policy choice. Higher cutoffs concentrate the training

**Question 3: Who was harmed, and by what mechanism (6 min)**  
"Amazon removed gender words and the model kept finding proxies. Name one proxy and
explain why it worked."  
Target: all-women's college names; gender-patterned extracurricular activities; writing style.  
Mechanism: any feature correlated with gender and with the historical hiring pattern
functions as a proxy, even without naming gender explicitly.

**Wrap-up line:**  
"Engineers tried to fix the problem at the Transform step by removing features.  
The problem was at the Represent step, in the labels.  
Patching a downstream operation does not fix an upstream error."

## Part 4: Fairness Audit Preview

Run the cells below during Segment 4 (Assignment Setup, 15 min) to show students
the kind of output they will produce in hw7.ipynb.

**Connection to the Amazon case:**  
The same proxy variable mechanism appears in this loan dataset.  
Zip code correlates with race. Including zip code as a model feature gives access to
race-correlated information even when race_category is removed from the feature set.
This is the loan-lending version of the resume-scoring proxy problem.

When framing this for students:  
"We traced this exact mechanism in the Amazon case tonight. Now you are going to find
it in a different domain. The pattern is the same. The dataset is different."

Dataset: 18 loan applicants, 9 columns.  
Path: `data.csv` (relative to Wednesday_Session/)

```python
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.3f}'.format)

df = pd.read_csv('data.csv')

print(f'Dataset loaded: {df.shape[0]} applicants, {df.shape[1]} columns')
print()
print('Columns:', list(df.columns))
print()
df.head(6)
```

```python
# HOMEWORK PREVIEW: hw7.ipynb Section 2
# Students compute approval rate by racial group and fill in a four-row table.

approval_by_race = (
    df.groupby('race_category')['approved'].mean() * 100
).round(1)

print('Approval rate by racial group (%):')
print(approval_by_race.to_string())
print()
print('(Students fill in the table in hw7.ipynb Section 2 using these numbers.)')
```

```python
# CONTEXT: Does credit score explain the approval gap?
# Show this if students ask during assignment setup.

credit_by_race = df.groupby('race_category')['credit_score'].mean().round(1)
n_by_race = df.groupby('race_category').size()

summary = pd.DataFrame({
    'approval_rate_pct': approval_by_race,
    'avg_credit_score':  credit_by_race,
    'n_applicants':      n_by_race
})

print('Fairness audit summary:')
summary
```

```python
# FILTER: Who is in the borderline credit score band?
# Helps answer: does the credit score gap fully explain the approval gap?

borderline = df[(df['credit_score'] >= 640) & (df['credit_score'] <= 700)]

print(f'Borderline applicants (credit score 640-700): {len(borderline)}')
print()
borderline[['applicant_id', 'credit_score', 'race_category', 'approved']]
```

```python
# HOMEWORK PREVIEW: hw7.ipynb Section 3
# The crosstab students interpret: zip code vs. race.
# Question: if we remove race_category but keep zip_code,
# does the model still have access to race-correlated information?

print('Applicants by zip code and race (crosstab):')
print(pd.crosstab(df['zip_code'], df['race_category']))
print()
print('Approval rate by zip code:')
print(df.groupby('zip_code')['approved'].mean().round(2))
```

## Part 4b: Section 4 Preview, The Missing Data Problem

Section 4 is new this term: no blanks, no template. Students write the whole
filter-group-summarize pipeline themselves. Run this if a student asks what the
output should look like, or if Part A discussion stalls and you want a second
concrete example of the Represent-to-exclusion mechanism from tonight's case.

**Connection to the Amazon case:** Amazon's engineers could not audit the model's
fairness toward excluded applicants because excluded applicants generate no outcome
data to audit. The same thing happens here: zero Black applicants were approved, so
there is no default-rate row to check for them. The exclusion happens so early that
even an error-rate audit has nothing left to measure.

```python
# HOMEWORK PREVIEW: hw7.ipynb Section 4 (students write this unscaffolded)
# Filter to approved applicants, then group by race, then mean the default rate.

approved = df[df['approved'] == 1]
result = approved.groupby('race_category')['default_rate'].mean().round(3)
print('Default rate among approved applicants, by race:')
print(result)
print()
print('Black applicants approved:',
      df[(df['race_category'] == 'Black') & (df['approved'] == 1)].shape[0])
```

```python
# FULL AUDIT SUMMARY: instructor quick-reference during assignment setup

print('--- Full Fairness Audit Summary ---')
print()
print('Approval rate by group (%):')
print(df.groupby('race_category')['approved'].mean().mul(100).round(1).to_string())
print()
print('Average credit score by group:')
print(df.groupby('race_category')['credit_score'].mean().round(1).to_string())
print()
print('Count by group:')
print(df.groupby('race_category').size().to_string())
```

## Connecting the Case to the Homework

Both cases share a structural feature with hw7.ipynb:

**Amazon + Section 3:** a proxy variable (zip code in the loan case; college name in the
Amazon case) carries group-correlated information even after the protected characteristic
is removed from the feature set. This is the mechanism students identify in Section 3 and
explain in Question 5c.

**COMPAS + Section 4:** this is the closer parallel. COMPAS's entire finding only exists
because someone disaggregated an outcome by group AND by error type, exactly what
Section 4 asks students to do (filter to approved applicants, then group by race, then
check default rate). Section 4's finding, that Black applicants vanish from the table
entirely because none were approved, is the same category of discovery as ProPublica's:
an aggregate number that looks fine on its own, until you split it by group and it isn't.
